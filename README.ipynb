# AI Eye Tracking Mouse 👁️🖱️

An AI-based mouse control system using computer vision, face tracking, and eye detection.

This project allows users to control the mouse cursor with head movement and perform clicks using eye blinking.

## 🚀 Features

- Face detection using OpenCV
- Eye detection
- Mouse control with head movement
- Smooth mouse movement
- Blink detection for mouse click

## 🛠️ Technologies

- Python
- OpenCV
- NumPy
- PyAutoGUI
